# Exploring Unsupervised Learning with K-Means Clustering
**Objective:** Explore unsupervised learning and clustering using K-Means with an existing dataset.

## Dataset used
We use the built-in **Iris dataset** from Scikit-learn. It contains measurements of 150 flowers across four features:
- Sepal length
- Sepal width
- Petal length
- Petal width

The dataset includes species labels, but **we will not use those labels to train K-Means**. This keeps the task unsupervised.

**How to use:** Upload this notebook to Google Colab and run cells from top to bottom. No separate dataset download is needed.

## 1. Import libraries
`pandas` helps display the data, `matplotlib` draws graphs, and Scikit-learn provides the existing dataset and K-Means algorithm.

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## 2. Load the existing Iris dataset
We load the data directly from Scikit-learn. The feature columns are used for clustering. The species column is kept only for reference and is not used to fit the model.

```python
# Load the built-in Iris dataset
iris = load_iris()

# Create a table containing the four numeric features
X = pd.DataFrame(iris.data, columns=iris.feature_names)

# Keep species names only for optional reference (not used by K-Means)
species_names = [iris.target_names[i] for i in iris.target]

print("Dataset size:", X.shape)
print("\\nFirst five rows:")
display(X.head())
print("\\nFeature names:", list(X.columns))
```

## 3. Prepare the data
K-Means uses distances, so features with larger numeric scales can dominate. We standardize each feature so it has approximately mean 0 and standard deviation 1.

```python
# Scale features before clustering
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled data shape:", X_scaled.shape)
```

## 4. Apply K-Means clustering
Choose `k = 3` to group the flowers into three clusters. K-Means learns from the four measurements only; it does not see the species labels.

```python
# Set the number of clusters
k = 3

# Create and train the K-Means model
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)

print("First 15 cluster assignments:", cluster_labels[:15])
print("Inertia:", round(kmeans.inertia_, 2))
```

## 5. Visualize the clusters using PCA
The Iris dataset has four features, so it cannot be plotted directly on a 2D graph. PCA projects the scaled data to two dimensions for visualization only. The K-Means clustering itself was performed using all four features.

```python
# Reduce the four features to two components for plotting
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 5))

# Color each flower by the cluster assigned by K-Means
plt.scatter(
    X_2d[:, 0], X_2d[:, 1],
    c=cluster_labels, cmap="viridis", s=50
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clusters on Iris Dataset (PCA View)")
plt.grid(True)
plt.show()

print("Variance shown by the two PCA components:",
      f"{pca.explained_variance_ratio_.sum():.1%}")
```

## 6. View cluster sizes
Cluster IDs (0, 1, 2) are arbitrary identifiers; they do not automatically correspond to species names.

```python
# Count how many flowers were assigned to each cluster
cluster_counts = pd.Series(cluster_labels).value_counts().sort_index()

for cluster_id, count in cluster_counts.items():
    print(f"Cluster {cluster_id}: {count} flowers")
```

## 7. Use the Elbow Method to explore k
The Elbow Method compares inertia for different values of `k`. Look for a point where the decrease in inertia begins to slow down. This is a guide, not a definitive rule.

```python
inertias = []
k_values = range(1, 8)

for number_of_clusters in k_values:
    model = KMeans(
        n_clusters=number_of_clusters,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    inertias.append(model.inertia_)

plt.figure(figsize=(7, 5))
plt.plot(list(k_values), inertias, marker="o")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method on Iris Dataset")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

## 8. Key takeaways
- The Iris dataset is an existing dataset available in Scikit-learn.
- Unsupervised learning finds groups without using target labels.
- K-Means groups observations based on their distances from cluster centers.
- Feature scaling is important when features have different scales.
- PCA is used here only to display four-dimensional data on a 2D graph.
- The Elbow Method helps explore a suitable value of `k`.
