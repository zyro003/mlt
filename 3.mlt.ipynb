# Implementing Linear Regression from Scratch
**Objective:** Understand linear regression by implementing it using basic Python and NumPy, without using a ready-made regression model.

## What this practical does
1. Creates a small sample dataset.
2. Visualizes the data.
3. Implements the prediction equation.
4. Calculates Mean Squared Error (MSE).
5. Learns the best slope and intercept using gradient descent.
6. Plots the fitted line and checks predictions.

**How to use:** Upload this `.ipynb` file to Google Colab and run the cells from top to bottom.

## 1. Import libraries
We use NumPy for calculations and Matplotlib for plotting. We will **not** use `LinearRegression` from scikit-learn because the goal is to implement the algorithm ourselves.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Create a simple dataset
Here, `X` represents study hours and `y` represents marks. The values are illustrative sample data.

```python
# Input feature: number of hours studied
X = np.array([1, 2, 3, 4, 5], dtype=float)

# Target/output: marks scored
y = np.array([35, 40, 50, 55, 65], dtype=float)

# Plot the original observations
plt.scatter(X, y)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.grid(True)
plt.show()
```

## 3. Understand the linear regression equation
A straight line is represented by:

\[
\hat{y} = mx + b
\]

- `x` = input (study hours)
- `ŷ` = predicted marks
- `m` = slope (how much the prediction changes when x increases by 1)
- `b` = intercept (predicted value when x is 0)

The model starts with initial values for `m` and `b`, then improves them using gradient descent.

```python
# Start with simple initial values
m = 0.0  # slope
b = 0.0  # intercept

# Predict output using the current slope and intercept
def predict(X, m, b):
    return m * X + b

print("Initial predictions:", predict(X, m, b))
```

## 4. Calculate Mean Squared Error (MSE)
MSE measures how far predictions are from actual values on average, with larger errors receiving more weight. A lower MSE means the predictions fit the training data better.

```python
def mean_squared_error(y_actual, y_predicted):
    # Find the squared difference for each observation, then take the mean
    return np.mean((y_actual - y_predicted) ** 2)

initial_predictions = predict(X, m, b)
print("MSE before training:", mean_squared_error(y, initial_predictions))
```

## 5. Train the model using gradient descent
Gradient descent updates the slope and intercept little by little to reduce MSE.

For this simple one-feature model, the gradients of the MSE are:
- Slope gradient: `(-2/n) * sum(X * (y - prediction))`
- Intercept gradient: `(-2/n) * sum(y - prediction)`

The learning rate controls the size of each update. If it is too large, training may become unstable; if too small, learning may be slow.

```python
# Training settings
learning_rate = 0.01
epochs = 5000  # Number of times the model updates its parameters
n = len(X)

# Store the error at each epoch so we can plot learning progress
loss_history = []

for epoch in range(epochs):
    # Step 1: Predict marks using current parameters
    y_pred = predict(X, m, b)

    # Step 2: Calculate errors (actual - predicted)
    errors = y - y_pred

    # Step 3: Calculate gradients for slope and intercept
    dm = (-2 / n) * np.sum(X * errors)
    db = (-2 / n) * np.sum(errors)

    # Step 4: Update parameters to reduce the error
    m = m - learning_rate * dm
    b = b - learning_rate * db

    # Step 5: Save MSE after the update
    loss_history.append(mean_squared_error(y, predict(X, m, b)))

print("Learned slope (m):", round(m, 3))
print("Learned intercept (b):", round(b, 3))
print("MSE after training:", round(loss_history[-1], 3))
```

## 6. Plot the loss curve
The loss curve shows how MSE changes during training. Ideally, the loss decreases and then becomes nearly stable.

```python
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.title("Gradient Descent: Loss Curve")
plt.grid(True)
plt.show()
```

## 7. Plot the fitted regression line
The points are the actual data. The line shows the values predicted by the trained model.

```python
# Calculate predicted marks for the dataset
y_final = predict(X, m, b)

plt.scatter(X, y, label="Actual data")
plt.plot(X, y_final, label="Regression line")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Linear Regression from Scratch")
plt.legend()
plt.grid(True)
plt.show()
```

## 8. Predict marks for a new input
Use the trained slope and intercept to predict marks for a student who studies for 6 hours. This is an illustrative prediction, not a guarantee of real exam performance.

```python
new_hours = 6
predicted_marks = predict(new_hours, m, b)

print(f"Predicted marks for {new_hours} study hours: {predicted_marks:.2f}")
```

## 9. Summary
- Linear regression models the relationship between an input and a numerical output with a straight line.
- The slope (`m`) and intercept (`b`) define the line.
- MSE measures prediction error.
- Gradient descent repeatedly adjusts `m` and `b` to reduce MSE.
- Once trained, the model can predict an output for a new input.
